```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep= ';', header= None,names= ["text",'emotion'])
```

```python
emotion_map = {
    'sadness': 1,
    'anger': 2,
    'love': 3,
    'surprise': 4,
    'fear': 5,
    'joy': 6
}

df['emotion'] = df['emotion'].map(emotion_map)
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
df
```

```python
def remove_number(txt):
    new = ""

    for i in txt:
        if not i.isdigit():
            new = new + i

    return new

df['text'] = df['text'].apply(remove_number)
df
```

```python
def remove_emoji(txt):
    new = ""

    for i in txt:
        if i.isascii():
            new = new + i

    return new

df["text"] = df["text"].apply(remove_emoji)

df
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.tokenize import word_tokenize
```

```python
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
stop_word = set(stopwords.words('english'))
```

```python
stop_word
```

```python
def remove_stopwords(txt):
    words = txt.split()
    new = []

    for word in words:
        if word.lower() not in stop_word:
            new.append(word)

    return " ".join(new)
```

```python
df["text"] = df["text"].apply(remove_stopwords)
```

```python
df
```

```python
X = df.drop("emotion",axis = 1)
y = df['emotion']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

```python
X_vector = CountVectorizer()
```

```python
X_train_bow = X_vector.fit_transform(X_train["text"])
X_test_bow = X_vector.transform(X_test["text"])
```

```python
from sklearn.naive_bayes import MultinomialNB

m_NB = MultinomialNB()
```

```python
m_NB.fit(X_train_bow,y_train)
```

```python
y_pred = m_NB.predict(X_test_bow)
```

```python
from sklearn.metrics import accuracy_score,classification_report

acc = accuracy_score(y_test,y_pred)
acc
```

```python
from sklearn.linear_model import LogisticRegression

le = LogisticRegression()
```

```python
le.fit(X_train_bow,y_train)
```

```python
y_le_pred = le.predict(X_test_bow)
```

```python
accsc = accuracy_score(y_test,y_le_pred)
```

```python
print(accsc)
print(classification_report(y_test,y_le_pred))
```

# LogisticRegression model is perform better in this DataSet

```python
tfidf = TfidfVectorizer()
```

```python
X_train_tfidf = tfidf.fit_transform(X_train["text"])
X_test_tfidf = tfidf.transform(X_test["text"])
```
